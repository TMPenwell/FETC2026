```python
import torch
```

```python
# Create a list of 5 test scores

scores_list = [85, 92, 78, 90, 88]
```

```python
# Convert the list to a PyTorch Tensor

my_tensor = torch.tensor(scores_list)
```

```python
print(f"Tensor Data: {my_tensor}")
```

```python
# Check the attributes of our tensor

print(f"Data Type: {my_tensor.dtype}")

print(f"Current Device: {my_tensor.device}")
```
